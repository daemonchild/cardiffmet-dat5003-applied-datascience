## Feature Engineering

As part of the exploration and preprocessing process, here I will use existing data to generate a number of useful features to allow further processing and analysis of the data set.


### Dates, Times and Days
From the 'Date' and 'Time' fields, it is simple to create a combined 'Date-and-Time' ('Datetime') feature. It is equally simple to split out the 'Time' field into a simpler 'Hour' feature.

- Datetime (combined)
- Hour of the day (24 hour clock)
- Day of the week (and day number)
- Month (and month number)

The weekday and months of the year require a custom ordering to be created, and they hjave been converted to the categorial data type.

```python
# Create 'Datetime'
df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'])

# Create 'Hour' by splitting the Time field
# For example, 13:00:00 -> 13
df['Hour'] = df['Time'].apply(lambda x: int(x.split(':')[0]))

# Create 'Weekday' and 'WeekdayNo'
df['Weekday'] = df['Datetime'].dt.day_name()
df['WeekdayNo'] = df['Datetime'].dt.dayofweek

# Create 'Weekday' and 'WeekdayNo'
df['Month'] = df['Datetime'].dt.month_name()
df['MonthNo'] = df['Datetime'].dt.month

# Create a custom order for the days of the week, or it will be alphabetical
weekday_custom_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
month_custom_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']

# Convert 'Weekday' to a categorical type uasing the custom order
df['Weekday'] = pd.Categorical(df['Weekday'], categories=weekday_custom_order, ordered=True)
df['Month'] = pd.Categorical(df['Month'], categories=month_custom_order, ordered=True)

# Check the data types for the new features:
# (df.dtypes creatres a series object, so need to recombine with the column names)

print (f"{Fore.YELLOW}[EDA]{Fore.RESET} Data types for the new features:")

dtypes_table = [(column, dtype) for column, dtype in zip(df.columns, df.dtypes)]
print(tabulate(dtypes_table, headers=["Feature", "Data Type"], tablefmt="grid"))
```

The following function draws multiple graphs in one large plot.

```python
def plot_graph_grid_line(df, fields_to_plot, x_field, graphs_per_row=3, y_label = ""):

    # Calculate the number of rows needed for the subplots
    _no_fields_to_plot = len(fields_to_plot)
    _graph_rows = math.ceil( (_no_fields_to_plot / graphs_per_row))

    # Create new plot grid (share x-axis for space, no shared y-axis to allow different scales)
    fig, axs = plt.subplots(_graph_rows,graphs_per_row, figsize=(18, 18), sharex=True, sharey=False, constrained_layout=True)

    # Set the starting location ot the top left corner of the grid
    _location_x = _location_y = 0

    # Loop through the fields to create subplots
    for _field in fields_to_plot:

        print (f"{Fore.YELLOW}[PLOT]{Fore.RESET} Plotting field: {_field}")

        # Make a new filtered dataframe 
        plot_df = df[[x_field, _field]]

        # Replace -200 with NaN and drop NaN values
        plot_df[_field].replace(-200, np.nan, inplace=True)
        #plot_df.dropna(inplace=True)

        # Plot the chart at the correct location in the grid
        plot_df.plot(kind='line',x=x_field, y=_field, ax=axs[_location_x, _location_y], subplots=True, legend=False, linewidth=0.5, color='#25a983')

        # Labels
        axs[_location_x, _location_y].set_title(_field)
        axs[_location_x, _location_y].set_xlabel(x_field)
        axs[_location_x, _location_y].set_ylabel(y_label)

        # Move to the next location in the grid
        if _location_y < (graphs_per_row - 1):
            _location_y += 1
        else:
            _location_y = 0
            _location_x += 1
```

I am going to use the K Nearest Neighbours algorithm to impute the missing values from the surrounding data.

Firstly, I am going to use the GridSearchCV (Grid Search with Cross Validation) algorithm to determine the best value for the number of neighbours to use. I will use KFold for cross validation. 

(https://scikit-learn.org/stable/modules/impute.html)

```python
# The KNNImputer function will fill NULL values, so the -200 must be replaced
df.replace(-200, np.nan, inplace=True)

no_rows = df.shape[0]
df.isna().sum().sort_values(ascending=False).apply(lambda x: f"{Fore.CYAN}{x}{Fore.RESET} -> {Fore.YELLOW}{round(x/no_rows*100, 1)}%{Fore.RESET}")
```

```python
# Optimise the value for the number of neighbours
# https://www.geeksforgeeks.org/how-to-find-the-optimal-value-of-k-in-knn/


# Define the fields to work with
impute_these_fields = gt_fields + pt08_fields + atmos_fields

# Define the range of k values to test
k_values = {'n_neighbors': list(range(1, 21))}

print (f"{Fore.CYAN}K_Values: {Fore.RESET}{k_values}")

# Initialize KNNImputer
knn_imputer = KNNImputer()

# Perform grid search with cross-validation
grid_search = GridSearchCV(knn_imputer, param_grid=k_values, cv=KFold(n_splits=5), scoring='neg_mean_squared_error')
grid_search.fit(df[impute_these_fields])

# Best k value
best_k = grid_search.best_params_['n_neighbors']
print(f"{Fore.CYAN}Optimal number of neighbors: {Fore.RESET} {best_k}")
```

```python
# Run the KNNImputer with the best k value
neighbours = best_k

# Create a new KNNImputer with the optimal number of neighbours
imputer = KNNImputer(n_neighbors=neighbours, weights="uniform")
imputer.fit_transform(df[impute_these_fields])
df[impute_these_fields] = imputer.fit_transform(df[impute_these_fields])
```

## Averaged Data - Monthly, Weekly and Daily

The following charts show the monthly, weekly and daily averages for normalised data. Normalising the data means that these features can be shown relative to one another on the same graph. This is necessary as the concentrations vary wildly in absolute terms.

The following function handles plotting the resampled data:

```python
def plot_multi_line_graph(df, fields_to_plot, resample="M", title=""):
  
    # Create new plot
    plt.figure(figsize=(8, 5))

    for field in fields_to_plot:

        # Resample to Monthly frequency and calculate the mean
        average = df[field].resample(resample).mean()

        average.plot(kind='line')
        plt.xlabel('Date')
        plt.ylabel('Relative Concentration')

    plt.legend(fields_to_plot, loc='upper right', bbox_to_anchor=(1.30, 1))
    plt.title(title, fontsize=14)
    plt.show()    
```

The following code filters the wanted fields from the dataset, normalises them and plots the graphs using the function above.

```python
# Filter the dataframe
fields_to_plot = gt_fields + pt08_fields
plot_df = df[fields_to_plot + ['Datetime']]

# Set Idatetime as index
plot_df.set_index('Datetime', inplace=True)

# Normalize the data fields
scaler = MinMaxScaler()
plot_df[fields_to_plot] = scaler.fit_transform(plot_df[fields_to_plot])

# Use the function above to plot the graphs
plot_multi_line_graph(plot_df, gt_fields, resample="D", title="Daily Average, Normalised Data")
plot_multi_line_graph(plot_df, pt08_fields, resample="D", title="Daily Average, Normalised Data")
plot_multi_line_graph(plot_df, gt_fields, resample="W", title="Weekly Average, Normalised Data")
plot_multi_line_graph(plot_df, pt08_fields, resample="W", title="Weekly Average, Normalised Data")
plot_multi_line_graph(plot_df, gt_fields, resample="M", title="Monthly Average, Normalised Data")
plot_multi_line_graph(plot_df, pt08_fields, resample="M", title="Monthly Average, Normalised Data")
```

## Have the Pollutant Levels Ever Exceeded Legal Safety Limits?

Is there a problem with the levels of pollutants in the air around the monitoring station? It is difficult to find standardised legal limits for each of the chemicals in the dataset. The data was gathered twenty years ago in 2004-2005. The World Health Organisation has proposed stricter limits since this time, but the legal limits expected by countries vary wildly. For instance, Russia allows an average of 100 microg/m<sup>3</sup>/m<sup>3</sup> of benzene (C<sub>6</sub>H<sub>6</sub>) over 24 hours, while the EU recommends only 3.25 microg/m<sup>3</sup> in the same time period. 

I have been able to find sensible values for safety limits for the following chemicals, which I will use to compare against the data to see how often these are breached. Again, it is important to note that the data was gathered 20 years ago, and safety in many countries is much more stringent now, while cars are subsequently much less polluting.

```python
# Define dictionary structure for maximum safe level averages
safe_level = {
                'CO':   {'1hr': 30.0, '8hr': 10.0, '24hr': 4.0},
                'C6H6': {'1hr': 35.0, '8hr': 3.25},
                'NO2':  {'1hr': 200.0, '24hr': 25.0}
            }

# Filter the dataframe
plot_df = df[['Datetime'] + gt_fields]
plot_df.set_index('Datetime', inplace=True)

total_slots = plot_df.shape[0]
eight_hour_slots = total_slots / 3
daily_slots = total_slots / 24

# Loop through the pollutants for which we have a 'safe level' value available
for chemical in safe_level:

    feature = chemical+"(GT)"

    # Set the number of times exceeded to 'unknown'
    number_exceeded_daily = number_exceeded_8hourly = number_exceeded_hourly = -1
    exceeded_dates_daily = []

    # If we have a 'save level' value, work on it.
    # Each is slightly different, so no further loop...

    if '24hr' in safe_level[chemical].keys():
        # Get a 24 hour average
        daily_average = plot_df[feature].resample('D').mean()
        number_exceeded_daily = (daily_average > safe_level[chemical]['24hr']).sum()  
        exceeded_dates_daily = daily_average[daily_average > safe_level[chemical]['24hr']].index

    if '8hr' in safe_level[chemical].keys():
        # Get a 8 hour average
        eighthourly_average = plot_df[feature].resample('8h').mean()
        number_exceeded_8hourly = (eighthourly_average > safe_level[chemical]['8hr']).sum()

    if '1hr' in safe_level[chemical].keys():
        # The values in the dataset are already averaged to one hour
        number_exceeded_hourly = (plot_df[feature] > safe_level[chemical]['1hr']).sum()

    # Print the results
    print (f"{Fore.YELLOW}--- {feature} --- {Fore.RESET}")

    print (f"{Fore.CYAN}Exceeded 24 hour average:{Fore.RESET}\t", end='')
    if number_exceeded_daily != -1:
        print (f"{number_exceeded_daily} times ({round ((number_exceeded_daily / daily_slots) * 100) }%)")
    else:
        print ("Safe Level Unknown")
        
    if len(exceeded_dates_daily) > 0:
        print (f"{Fore.CYAN} - On the following dates:{Fore.RESET}")
        print (exceeded_dates_daily)
        print ()

    print (f"{Fore.CYAN}Exceeded 8 hour average:{Fore.RESET}\t", end='')
    if number_exceeded_8hourly != -1:
        print (f"{number_exceeded_8hourly} times ({round ((number_exceeded_8hourly / eight_hour_slots) * 100) }%)")
    else:
        print ("Safe Level Unknown")

    print (f"{Fore.CYAN}Exceeded 1 hour average:{Fore.RESET}\t", end='')
    if number_exceeded_hourly != -1:
        print (f"{number_exceeded_hourly} times ({round ((number_exceeded_hourly / total_slots) * 100) }%)")
    else:
        print ("Safe Level Unknown")

    print ()
```

The following function is similar to previous 'multi-graph' functions user previously, but for heat maps.

```python
def plot_graph_grid_heatmap(df, fields_to_plot, row_field, column_field, graphs_per_row=3):

    # Calculate the number of rows needed for the subplots
    _no_fields_to_plot = len(fields_to_plot)
    _graph_rows = math.ceil( (_no_fields_to_plot / graphs_per_row))

    # Create new plot grid (shared y-axis as scale is comparible)
    fig, axs = plt.subplots(_graph_rows,graphs_per_row, figsize=(18, 18), sharex=True, sharey=True, constrained_layout=True)

    # Set the starting location ot the top left corner of the grid
    _location_x = _location_y = 0

    # Loop through the fields to create subplots
    for _field in fields_to_plot:

        print (f"{Fore.YELLOW}[PLOT]{Fore.RESET} Plotting field: {_field}")

        # Pivot the data to get hours as rows and weekdays as columns
        pivot_table = plot_df.pivot_table(values=_field, index=row_field, columns=column_field, aggfunc='mean')

        # Annot set to False as the colour gives more impact than the number 
        sns.heatmap(pivot_table, annot=False, cbar=False, cmap='viridis',ax=axs[_location_x, _location_y])

        # Labels
        axs[_location_x, _location_y].set_title(_field)
        axs[_location_x, _location_y].set_ylabel(row_field)
        axs[_location_x, _location_y].set_xlabel(column_field)

        # Move to the next location in the grid
        if _location_y < (graphs_per_row - 1):
            _location_y += 1
        else:
            _location_y = 0
            _location_x += 1


    plt.show();
```
